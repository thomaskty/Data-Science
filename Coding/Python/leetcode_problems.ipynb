## 3753. Total Waviness of Numbers in Range II
- bruteforce 

```python
## number of peaks and valleys for a number:
# window of 3 

def get_peaks_valleys(n):
    nums = list(map(int, str(n)))
    peaks = 0 
    valleys = 0
    if len(nums)<3:
        return peaks,valleys
    else:
        for i in range(1,len(nums)-1):
            if nums[i-1]>nums[i]<nums[i+1]:
                print(f'vallyes = {[nums[i-1],nums[i],nums[i+1]]}')
                valleys+=1
            elif nums[i-1]<nums[i]>nums[i+1]:
                print(f'peaks = {[nums[i-1],nums[i],nums[i+1]]}')
                peaks+=1
        return peaks+valleys
```

```python
# bruteforce solution : first try : 
def totalWaviness(self, num1: int, num2: int) -> int:
    ans = 0
    def get_peaks_valleys(n):
        nums = list(map(int, str(n)))
        peaks = 0 
        valleys = 0
        if len(nums)<3:
            return peaks+valleys
        else:
            for i in range(1,len(nums)-1):
                if nums[i-1]>nums[i]<nums[i+1]:
                    # print(f'vallyes = {[nums[i-1],nums[i],nums[i+1]]}')
                    valleys+=1
                elif nums[i-1]<nums[i]>nums[i+1]:
                    # print(f'peaks = {[nums[i-1],nums[i],nums[i+1]]}')
                    peaks+=1
            return peaks+valleys
    for x in range(num1, num2+1):
        ans += get_peaks_valleys(x)
    return ans 
```

## 242 : Valid Anagrams

```python
def is_anagram(s,t):
    # if length mismatch then return false 
    if len(s)!=len(t):
        return False
    
    # getting the counts 
    ds,dt = {},{}
    for i in range(len(s)):
        ds[s[i]] = ds.get(s[i],0)+1
        dt[t[i]] = dt.get(t[i],0)+1

    for i in ds:
        if ds[i] != dt.get(i,0):
            return False
    return True
```

```python
s = 'anagram'
t = 'nagaram'

is_anagram(s,t)
```

```python
s = 'rat'
t = 'car'
is_anagram(s,t)
```
