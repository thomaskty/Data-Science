```python
def argsort(arr):
    n = len(arr)
    idx = [i for i in range(n)]
    wall = n - 1
    while wall > 0:
        for i in range(wall):
            if arr[idx[i]] > arr[idx[i + 1]]:
                idx[i], idx[i + 1] = idx[i + 1], idx[i]
        wall -= 1
    return idx

argsort([4,2,8,5,23,])
```
