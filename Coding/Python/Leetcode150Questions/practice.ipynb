```python
def transpose(matrix):
    # sample = [[2,4,5,0],[4,3,2,5]]
    # output = [[2,4][4,3],[5,2],[0,5]]
    n_rows = len(matrix)
    n_cols = len(matrix[0])
    transposed = [[0]*n_rows for i in range(n_cols)]
    for i in range(n_rows):
        for j in range(n_cols):
            transposed[j][i] = matrix[i][j]
    return transposed
```

```python
def sorting(ar):
    wall = len(ar)-1
    while wall>0:
        for i in range(wall):
            if ar[i]>ar[wall]:
                ar[i],ar[wall] = ar[wall],ar[i]
        wall-=1
    return ar
```

```python
def count_dict(ar):
    counts = {}
    for i in ar:
        if i not in counts:
            counts[i] = 1
        else:
            counts[i]+=1
    return counts
```

```python
def reverse(ar):
    i,j = 0,len(ar)-1
    while i<j:
        ar[i],ar[j] = ar[j],ar[i]
        i+=1
        j-=1
    return ar
```

```python
def rotate_anticlockwise(ar):
    for i in range(len(ar)-1,-1,-1):
        ar[len(ar)-1],ar[i] = ar[i],ar[len(ar)-1]
    return ar
```

```python
def rotate_k_forward(ar,k):
    n = len(ar)
    k = k%n
    s = n-k
    return ar[s:]+ar[:s]
```

```python
def rotate_k_times_inplace(ar, k):
    n = len(ar)
    k %= n
    def reverse(ar, i, j):
        while i < j:
            ar[i], ar[j] = ar[j], ar[i]
            i += 1
            j -= 1

    reverse(ar, 0, n-1)
    reverse(ar, 0, k-1)
    reverse(ar, k, n-1)

    return ar
```

```python
def prime(n):
    if n<=1:
        return False
    else:
        for i in range(2,n):
            if n%i==0:
                return False
    return True
```

```python
def factorial(n):
    if n==1:
        return 1
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
```

```python
def is_palindrome(string):
    _valid = list('qwertyuiopasdfghjklzxcvbnm')
    _valid+= list('QWERTYUIOPASDFGHJKLZXCVBNM')
    _valid+= list('0123456789')
    i,j = 0,len(string)-1
    while i<j:

        # skipping the left value if it is not in the _valid list
        while (i<j) and (string[i] not in _valid):
            i+=1

        # skipping the right value if it is not in the _valid list
        while (i<j) and (string[j] not in _valid):
            j-=1

        # comparing the left and right values
        if string[i].lower()==string[j].lower():
            i+=1
            j-=1
        else:
            return False

    return True
```

```python
def arg_sort(arr):
    n = len(arr)

    # list of indices
    idx = [i for i in range(n)]

    wall = n - 1
    while wall > 0:
        for i in range(wall):
            # compare values in arr using indices in idx
            if arr[idx[i]] > arr[idx[i + 1]]:
                idx[i], idx[i + 1] = idx[i + 1], idx[i]
        wall -= 1

    return idx
```

```python
def longest_word(sentence):
    max_length = 0
    start_index = 0      # start of current word
    longest_start = 0    # start of longest word found
    longest_end = 0      # end of longest word found

    for i in range(len(sentence)):
        if sentence[i] != ' ':
            # if it's the start of a word, mark start_index
            if i == 0 or sentence[i-1] == ' ':
                start_index = i

            # if end of sentence or next char is space, check word length
            if i == len(sentence) - 1 or sentence[i+1] == ' ':
                word_length = i - start_index + 1
                if word_length > max_length:
                    max_length = word_length
                    longest_start = start_index
                    longest_end = i

    return max_length, sentence[longest_start:longest_end+1]
```

```python
def missing_number(ar):
    missing = []
    for i in range(len(ar)-1):
        diff = ar[i+1]-ar[i]
        if diff>1:
            for j in range(ar[i]+1,ar[i+1]):
                missing.append(j)
        else:
            pass
    return missing
```

```python
def is_power_of_two(n: int) -> bool:
    if n < 1:
        return False  # 0 or negative numbers are not powers of two

    while n % 2 == 0:
        print(f'n = {n}')
        n = n//2  # divide n by 2 repeatedly

    return n == 1  # if we ended at 1, it was a power of two
```

```python
def max_subarray_sum(ar):
    max_subarray_val = float('-inf')
    for i in range(len(ar)):
        for j in range(i,len(ar)):
            sum_=0
            for k in range(i,j+1):
                sum_+=ar[k]

            if max_subarray_val<sum_:
                max_subarray_val = sum_
    return max_subarray_val
```

```python
def are_anagrams_count(s1: str, s2: str) -> bool:
    if len(s1) != len(s2):
        return False
    count1 = {}
    count2 = {}

    for ch in s1:
        count1[ch] = count1.get(ch, 0) + 1

    for ch in s2:
        count2[ch] = count2.get(ch, 0) + 1

    return count1 == count2
```

```python
def binary_search(ar,value,start,end):
    if start<=end:
        middle = (start + end) // 2
        if value == ar[middle]:
            return middle
        elif value>ar[middle]:
            return binary_search(ar,value,middle+1,end)
        elif value<ar[middle]:
            return binary_search(ar,value,start,middle-1)
        return None
    else:
        return -1
```

```python
# finding count of value in a sorted array
def find_occurrence(arr,value,first=True):
    low,high = 0,len(arr)-1
    result = -1
    while low<=high:
        mid = (low+high)//2
        if arr[mid] == value:
            result = mid
            if first:
                high = mid-1
            else:
                low = mid+1
        elif arr[mid]<value:
            low = mid+1
        else:
            high = mid-1
    return result

def find_count(ar,value):
    first = find_occurrence(ar,value,first = True)
    last = find_occurrence(ar,value,first =False)
    if first==-1:
        return -1
    return (last-first)+1
```

```python
def is_sorted_and_rotated(nums):
    N = len(nums)
    count = 1
    for i in range(1,2*N):
        if nums[(i-1)%N] <=nums[i%N]:
            count+=1
        else:
            count = 1
        if count ==N:
            return True
    return N==1
```

```python
# index of a value in a circularly sorted array with no duplicates
def circular_search(ar,n,x):
    low = 0; high = n-1
    while low<=high:
        mid = int((low+high)/2)
        if x==ar[mid]:  # case 1 : found x
            return mid
        if ar[mid]<=ar[high]: # case 2 ; right half is sorted
            if ar[mid] < x <= ar[high]:
                low = mid +1
            else:
                high = mid -1
        else:
            if ar[low] <= x < ar[mid]:
                high = mid -1
            else:
                low = mid+1
    return -1
```

```python
def two_sum(numbers, target): # o(n)
    h = {}
    for i, num in enumerate(numbers):
        desired = target - num
        if desired in h:
            # return num,desired # returning the numbers
            return h[desired],i  # returning the indexes of numbers
        h[num] = i
    return None
```

```python
def two_sum_sorted(ar,target):
    i,j = 0,len(ar)-1
    while i<j:
        _sum = ar[i]+ ar[j]
        if _sum <target:
            i+=1
        elif _sum>target:
            j-=1
        else:
            return i+1,j+1
```
