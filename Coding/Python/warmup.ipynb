```python
# question 1 
def find_equilibrium_indices(nums):
    e_indices = []
    n = len(nums)
    for i in range(n):
        
        # find the left sum 
        left_sum = 0
        if (i+1)<n:
            left_array = nums[i+1:]
            for value in left_array:
                left_sum +=value
        else:
            left_sum = 0 

        # find right sum 
        right_sum = 0
        right_array = nums[:i]
        if not right_array:
            pass
        else:
            for value in right_array:
                right_sum+=value

        # compare
        if left_sum==right_sum:
            e_indices.append(i) 
    
    return e_indices

# question 2 
def find_triplets(nums,target):
    s = dict()
    for i in range(len(nums)):
        s[nums[i]] = i 
        for j in range(i+1,len(nums)):
            s[nums[j]] = j 
            diff = target-(nums[j]+nums[i])
            if diff in s:
                return i,j,s[diff]
    return None 

# question 3 
def merge_two_sorted_arrays(nums1,nums2):
    i,j = len(nums1)-1,len(nums2)-1 # last elements 
    output = [0]*(i+j) 
    p = i+j-1
    while i>0 and j>0:
        if nums1[i]>nums2[j]:
            output[p]= nums1[i]
            i-=1
            p-=1
            print(f'replacing with nums1 : {output}')
        else:
            output[p] = nums2[j]
            j-=1
            p-=1
            print(f'replacing with nums2 : {output}')
            
    while i>0:
        output[p] = nums1[i]
        i-=1
        p-=1
        print(f'Extra values : replacing with nums1 : {output}')

    while j>0:
        output[p] = nums2[j]
        j-=1
        p-=1
        print(f'Extra values : replacing with nums1 : {output}')

    return output

# question 4 
def max_subarray(nums):
    max_i,max_j = 0,0
    max_sum = 0
    for i in range(len(nums)):
        for j in range(i,len(nums)):
            current_sum = 0
            for k in range(i,j+1):
                current_sum+=nums[k]
                if current_sum>max_sum:
                    max_sum = current_sum
                    max_i = i
                    max_j = j
                    
    return max_sum, nums[max_i:max_j+1]

# question 5 
def two_sum(nums,t):
    s = dict()
    for i in range(len(nums)):
        if t-nums[i] in s:
            return i,s[t-nums[i]]
        s[nums[i]] =  i 
    return s

# question 6 
def union_arrays(nums1,nums2):
    for i in nums1:
        nums2.append(i)
    return nums2     

# question 7 
def intersection_two_arrays(nums1,nums2):
    output = [] 
    for i in nums2:
        if i in nums1:
            output.append(i)
    return output    

# question 8 
def find_duplicates(nums):
    count_dict = {}
    for i in range(len(nums)):
        if nums[i] not in count_dict:
            count_dict[nums[i]] = 0 
        count_dict[nums[i]]+= 1
    return [i for i in count_dict if count_dict[i]>1]

# question 9 
def max_min(nums):
    max_ = nums[0]
    min_ = nums[0]
    for i in range(len(nums)):
        if nums[i]>max_:
            max_ = nums[i]
        if nums[i]<min_:
            min_ = nums[i]
    return max_,min_

# question 10 
def reverse_inplace(nums):
    i,j = 0,len(nums)-1
    while i<j:
        nums[i],nums[j] = nums[j],nums[i]
        i+=1;j-=1
    return nums
    
# question 11 
def rotate_one_step(nums):
    n = len(nums)
    for i in range(1,len(nums)):
        nums[0],nums[i] = nums[i],nums[0]
    return nums

# question 12 
def rotate_array_k_forward(nums,k):
    for _ in range(k):
        nums = rotate_one_step(nums)
    return nums

# question 13 
def move_zeros_end(nums):
    i,j = 0,len(nums)-1
    while i<j:
        while i<j and nums[i]!=0:
            i+=1
        while i<j and nums[j]==0:
            j-=1
        nums[i],nums[j] = nums[j],nums[i]
        i+=1;j-=1
    return nums

# question 14 
def is_sorted(nums,ascending= True):
    for i in range(len(nums)-1):
        if ascending:
            if nums[i]>nums[i+1]:
                return False
        else:
            if nums[i]<nums[i+1]:
                return False
    return True

# question 15 
def remove_duplicates_from_sorted_array(nums):
    k = 0
    i = 0 
    while i<len(nums):
        if i+1<len(nums) and nums[i] == nums[i+1]:
            i+=1
        else:
            nums[k] = nums[i]
            k+=1
            i+=1
    return nums[:k]

# question 16
def find_missing_nums_12n(nums):
    # assuming that  the numbers are ordered 
    missing = [] 
    for i in range(1,len(nums)):
        diff = nums[i]-nums[i-1]
        if diff>1:
            for j in range(nums[i-1]+1,nums[i]):
                missing.append(j)
        else:
            pass
    return missing   

# question 17 

def get_counts(nums):
    count = {} 
    for i in range(len(nums)):
        if nums[i] not in count:
            count[nums[i]] = 1
        else:
            count[nums[i]] +=1
    return count

# question 18 
def peak_element(nums):
    # assuming that we only have one peak element 
    count = {} 
    for i in range(len(nums)):
        if nums[i] not in count:
            count[nums[i]] = 1
        else:
            count[nums[i]] +=1

    max_count = 0
    peak_element = nums[0] 
    for i in count:
        if count[i]>max_count:
            max_count = count[i]
            peak_element = i
    return peak_element
```
