## ✔️ Converting Roman Numbers to Integer 

```python
def roman_to_integer(roman: str) -> int:
    # Map Roman numerals to their integer values
    roman_map = {
        'I': 1,
        'V': 5,
        'X': 10,
        'L': 50,
        'C': 100,
        'D': 500,
        'M': 1000
    }
    
    total = 0  # Result accumulator
    n = len(roman)
    
    for i in range(n):
        current_value = roman_map[roman[i]]
        
        # If this is not the last character and next numeral is larger, subtract
        if i < n - 1 and roman_map[roman[i]] < roman_map[roman[i + 1]]:
            total -= current_value
        else:
            total += current_value
    
    return total
```

## ✔️ Binary Search While & Recursive Implementation

```python
# binary search recursive implementation 
def binary_search(ar,value,start,end):
    if start<=end:
        middle = (start + end) // 2
        if value == ar[middle]: 
            return middle
        elif value>ar[middle]:  
            return binary_search(ar,value,middle+1,end)
        elif value<ar[middle]:
            return binary_search(ar,value,start,middle-1)
        return None
    else:
        return -1
```

```python
array = [3,4,5,6,18,32,98]
binary_search(array,32,0,len(array))
```

```python
# binary search while loop implementation  
def binary_search_while(ar,value):
    start,end = 0,len(ar)-1
    while start<=end:
        mid = (start + end) // 2
        if value == ar[mid]:
            return mid 
        elif value>ar[mid]:
            start = mid+1 
        elif value<ar[mid]:
            end = mid-1
    return -1
```

```python
array = [3,4,5,6,18,32,98]
binary_search_while(array,32)
```

## ✔️ Bubble Sort Implementation

```python
# O(n2)
def bubble_sort(arr):
    wall = len(arr)-1
    while wall > 0:
        for i in range(wall):
            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
        wall -= 1
    return arr
```

## ✔️ Find occurence of a value in a sorted array : using binary search 

```python
def find_occurrence(arr,value,first=True):
    low,high = 0,len(arr)-1
    result = -1
    while low<=high:
        mid = (low+high)//2
        if arr[mid] == value:
            result = mid
            if first:
                high = mid-1
            else:
                low = mid+1
        elif arr[mid]<value:
            low = mid+1
        else:
            high = mid-1
    return result
```

## ✔️ Finding count of value in a sorted array 

```python
def find_count(ar,value):
    first = find_occurrence(ar,value,first = True)
    last = find_occurrence(ar,value,first =False)
    if first==-1:
        return -1
    return (last-first)+1
```

## ✔️ Fibonacci Series

```python
def fibanocci_recursion(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fibanocci_recursion(n-1)+fibanocci_recursion(n-2)

# while loop implementation
def fibanocci_while(n):
    if n==0:
        return 0 
    elif n==1:
        return 1 
    else:
        a,b = 0,1
        i = 0
        while i< n-1:
            next_value = a+b 
            a,b = b,next_value
            i+=1 
        return b

# for loop implementation
def fib_iterative(n):
    if n == 0 or n == 1:
        return 1
    a, b = 1, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

# using memoization
def fib_memoization(n, memo={0: 1, 1: 1}):
    if n in memo:
        return memo[n]
    memo[n] = fib_memoization(n - 1, memo) + fib_memoization(n - 2, memo)
    return memo[n]
```

## 🅛 Search in Rotated Sorted Array: Leetcode 33
 * There is an integer array nums sorted in ascending order (with distinct values).
 * Prior to being passed to your function, nums is possibly left rotated at an unknown index k (1 <= k < nums.length) such that the resulting array is [nums[k], nums[k+1], ..., nums[n-1], nums[0], nums[1], ..., nums[k-1]] (0-indexed). For example, [0,1,2,4,5,6,7] might be left rotated by 3 indices and become [4,5,6,7,0,1,2].
 * Given the array nums after the possible rotation and an integer target, return the index of target if it is in nums, or -1 if it is not in nums.
 * You must write an algorithm with O(log n) runtime complexity.

```python
# index of a value in a circularly sorted array with no duplicates
def circular_search(ar,n,x):
    low = 0; high = n-1
    while low<=high:
        mid = int((low+high)/2)
        if x==ar[mid]:  # case 1 : found x
            return mid
        if ar[mid]<=ar[high]: # case 2 ; right half is sorted
            if ar[mid] < x <= ar[high]:
                low = mid +1
            else:
                high = mid -1
        else:
            if ar[low] <= x < ar[mid]:
                high = mid -1
            else:
                low = mid+1
    return -1
```

## 🅛 Jump Game : Leetcode 55
You are given an integer array nums. You are initially positioned at the array's first index, and each element in the array represents your maximum jump length at that position.
Return true if you can reach the last index, or false otherwise.

```python
# greedy backward solution

def canJump(nums):
    goal = len(nums)-1 # last element 
    for i in range(len(nums)-2,-1,-1):
        if i+nums[i]>=goal: # index + possible jump value is higher than goal 
            goal=i # shifting the goal to the current index ( one down ) 
    if goal==0:
        return True
    else:
        return False
```

## 🅛 Minimal Jump Required : Leetcode 45
The objective shifts from feasibility to minimum number of jumps required to reach the last index. A greedy “layered BFS” approach is optimal (O(n)). We assume that solution exists always. 

```python
def jump(nums):
    res = 0 
    l = r = 0 
    while r<len(nums)-1:
        farthest = 0
        for i in range(l,r+1):
            farthest = max(farthest,i+nums[i])
        l = r+1 
        r = farthest
        res +=1
    return res
```

## 🅛 Best Time to Buy and Sell Stock : Leetcode 121
 * You are given an array prices where prices[i] is the price of a given stock on the ith day.
 * You want to maximize your profit by choosing a single day to buy one stock and choosing a different day in the future to sell that stock.
 * Return the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return 0

```python
# prices = [7,1,5,3,6,4] 
# output = 5, buy at 1 and sell at 6

def maxProfit(nums):
    buy = nums[0]
    maxprofit = 0
    for i in range(0,len(nums)):
        if nums[i]<buy:
            buy = nums[i]

        profit = nums[i]-buy
        if profit>maxprofit:
            maxprofit = profit

    return maxprofit
```

```python
# Given an integer array nums, find the subarray with the largest sum, and return its sum
def maxsubarray_o3(nums):
    max_ = nums[0]
    max_i,max_j = 0,0 
    for i in range(len(nums)):
        for j in range(i,len(nums)):
            subarray_sum = 0 
            for k in range(i,j+1):
                subarray_sum+=nums[k]

            if subarray_sum>max_:
                max_ = subarray_sum
                max_i,max_j = i,j
    return max_,nums[max_i:max_j+1]
```

## ✔️ Majority Element : Leetcode 169
Given an array nums of size n, return the majority element. The majority element is the element that appears more than ⌊n / 2⌋ times. You may assume that the majority element always exists in the array.

```python
def majorityElement(nums):
    candidate = None
    count = 0
    for num in nums:
        if count == 0:
            candidate = num  # New candidate
            count = 1
            
        elif num == candidate:
            count += 1  # Same candidate, increment count
        else:
            count -= 1  # Different element, decrement count
    return candidate
```

## ✔️ Number of rotations happend in a sorted array with no duplicates.

```python
# rotations happend in a sorted array with no duplicates 
def get_rotations(ar,n):
    low,high = 0,n-1
    while low<=high:
        if ar[low]<=ar[high]:  # case 1 
            return low
        else:
            mid = int((low+high)/2)
            next_ = (mid+1)%n 
            prev = (mid+n-1)%n 
        if ar[mid]<=ar[next_] and ar[mid]<=ar[prev]: # case 2 
            return mid
        elif ar[mid]<=ar[high]:
            high = mid-1
        elif ar[mid] >=ar[low]:
            low = mid+1
    return -1
```

## 🅛 Valid Palindrome : Leetcode 125
A phrase is a palindrome if, after converting all uppercase letters into lowercase letters and removing all non-alphanumeric characters, it reads the same forward and backward. Alphanumeric characters include letters and numbers.

```python
# Time: O(n)
# Space: O(1)
def is_palindrome(s):
    left, right = 0, len(s) - 1

    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1

        if s[left].lower() != s[right].lower():
            return False

        left += 1
        right -= 1

    return True
```

## 🅛  Leetcode 1 : Two sum

```python
# Problem: Two Sum (Unsorted)
# Given an integer array nums and an integer target, return indices [i, j]
# such that nums[i] + nums[j] == target and i != j.
#
# Example 1:
# Input: nums = [2, 7, 11, 15], target = 9
# Output: [0, 1]
#
# Example 2:
# Input: nums = [3, 2, 4], target = 6
# Output: [1, 2]
#
# Constraints:
# 2 <= len(nums) <= 10^4
# -10^9 <= nums[i], target <= 10^9
# Exactly one valid answer may be assumed for classic version.

# Time: O(n^2)
# Space: O(1)
def two_sum_bruteforce(nums, target):
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []


# Time: O(n)
# Space: O(n)
def two_sum_hashmap(nums, target):
    seen = {}  # value -> index

    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i

    return []


# Problem: Two Sum II (Sorted Input)
# Given sorted nums, return indices [left, right] where nums[left] + nums[right] == target.
#
# Example:
# Input: nums = [2, 3, 4, 7, 11, 15], target = 9
# Output: [1, 2]
#
# Constraints:
# nums is sorted in non-decreasing order.
# 2 <= len(nums) <= 10^5

# Time: O(n)
# Space: O(1)
def two_sum_sorted(nums, target):
    left, right = 0, len(nums) - 1

    while left < right:
        total = nums[left] + nums[right]

        if total == target:
            return [left, right]
        if total < target:
            left += 1
        else:
            right -= 1

    return []
```

## 🅛  Leetcode 15 : Three sum 
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.

```python
# Problem: 3Sum
# Given integer array nums, return all unique triplets [a, b, c]
# such that a + b + c == 0.
#
# Example 1:
# Input: nums = [-1,0,1,2,-1,-4]
# Output: [[-1,-1,2],[-1,0,1]]
#
# Example 2:
# Input: nums = [0,1,1]
# Output: []
#
# Constraints:
# 3 <= len(nums) <= 3000
# -10^5 <= nums[i] <= 10^5

# Time: O(n^3)
# Space: O(k) for unique triplets output
def three_sum_bruteforce(nums):
    n = len(nums)
    unique = set()

    for i in range(n):
        for j in range(i + 1, n):
            for k in range(j + 1, n):
                if nums[i] + nums[j] + nums[k] == 0:
                    unique.add(tuple(sorted([nums[i], nums[j], nums[k]])))

    return [list(triplet) for triplet in unique]


# Time: O(n^2)
# Space: O(k) for unique triplets output
def three_sum_two_pointers(nums):
    nums.sort()
    result = []
    n = len(nums)

    for i in range(n - 2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        left, right = i + 1, n - 1

        while left < right:
            total = nums[i] + nums[left] + nums[right]

            if total == 0:
                result.append([nums[i], nums[left], nums[right]])

                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1

                left += 1
                right -= 1
            elif total < 0:
                left += 1
            else:
                right -= 1

    return result


# Problem: 3Sum Closest
# Return the sum of the triplet closest to target.
#
# Example:
# Input: nums = [-1,2,1,-4], target = 1
# Output: 2
#
# Constraints:
# 3 <= len(nums) <= 10^3
# -10^3 <= nums[i], target <= 10^3

# Time: O(n^2)
# Space: O(1)
def three_sum_closest(nums, target):
    nums.sort()
    closest = float("inf")

    for i in range(len(nums) - 2):
        left, right = i + 1, len(nums) - 1

        while left < right:
            current = nums[i] + nums[left] + nums[right]

            if abs(current - target) < abs(closest - target):
                closest = current

            if current < target:
                left += 1
            elif current > target:
                right -= 1
            else:
                return current

    return closest
```

## Move zeros to left 

```python
# Problem: Move Zeroes to End
# Move all zeroes in nums to the end while maintaining relative order of non-zero elements.
#
# Example 1:
# Input: nums = [0,1,0,3,12]
# Output: [1,3,12,0,0]
#
# Example 2:
# Input: nums = [0]
# Output: [0]
#
# Constraints:
# 1 <= len(nums) <= 10^4
# -2^31 <= nums[i] <= 2^31 - 1


# Time: O(n)
# Space: O(1)
def move_zeros_right_unstable(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        while left < right and arr[left] != 0:
            left += 1
        while left < right and arr[right] == 0:
            right -= 1

        if left < right:
            arr[left], arr[right] = arr[right], arr[left]

    return arr


# Time: O(n)
# Space: O(n)
def move_zeros_right_extra_space(arr):
    output = []

    for value in arr:
        if value != 0:
            output.append(value)

    while len(output) < len(arr):
        output.append(0)

    return output


# Time: O(n)
# Space: O(1)
def move_zeros_right_stable(arr):
    write = 0

    for read in range(len(arr)):
        if arr[read] != 0:
            arr[write] = arr[read]
            write += 1

    for i in range(write, len(arr)):
        arr[i] = 0

    return arr


test_cases = [
    ([0, 1, 0, 3, 12], [1, 3, 12, 0, 0]),
    ([1, 2, 3, 4], [1, 2, 3, 4]),
    ([0, 0, 0], [0, 0, 0]),
    ([0, 0, 1, 2, 3], [1, 2, 3, 0, 0]),
    ([4, 0, 5, 0, 0, 3, 0, 1], [4, 5, 3, 1, 0, 0, 0, 0]),
]
```

## Pivot Partition

```python
# Problem: Partition Array Around Pivot
# Reorder array so that:
# - elements < pivot appear first
# - elements == pivot appear next
# - elements > pivot appear last
#
# Example:
# Input: nums = [9,12,5,10,14,3,10], pivot = 10
# Output: [9,5,3,10,10,12,14] (any valid partition is accepted)
#
# Constraints:
# 1 <= len(nums) <= 10^5
# -10^9 <= nums[i], pivot <= 10^9

# Time: O(n)
# Space: O(n)
def pivot_array_extra_space(nums, pivot):
    less = []
    equal = []
    greater = []

    for value in nums:
        if value < pivot:
            less.append(value)
        elif value == pivot:
            equal.append(value)
        else:
            greater.append(value)

    return less + equal + greater


# Time: O(n)
# Space: O(1)
def pivot_array_dutch_flag(nums, pivot):
    low = 0
    mid = 0
    high = len(nums) - 1

    while mid <= high:
        if nums[mid] < pivot:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] > pivot:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
        else:
            mid += 1

    return nums
```

## Partition Odd & Even

```python
# Problem: Partition Odds and Evens
# Rearrange array in-place so all odd numbers appear before all even numbers.
# Relative order does not need to be preserved.
#
# Example:
# Input: [3,2,8,0,1,7,4]
# Output: [3,7,1,0,8,2,4] (any valid odd-left-even-right partition accepted)
#
# Constraints:
# 1 <= len(arr) <= 10^5
# -10^9 <= arr[i] <= 10^9

# Time: O(n)
# Space: O(1)
def partition_odd_even(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        while left < right and arr[left] % 2 == 1:
            left += 1
        while left < right and arr[right] % 2 == 0:
            right -= 1

        if left < right:
            arr[left], arr[right] = arr[right], arr[left]

    return arr
```

## K - Sum 

```python
# Problem: k-Sum (Generalization of 2Sum / 3Sum / 4Sum)
# Return all unique k-sized subsets from nums that sum to target.
#
# Example:
# Input: nums = [1,0,-1,0,-2,2], target = 0, k = 4
# Output: [[-2,-1,1,2],[-2,0,0,2],[-1,0,0,1]]
#
# Constraints (typical interview bounds):
# 1 <= len(nums) <= 200
# -10^9 <= nums[i], target <= 10^9
# 2 <= k <= len(nums)

# Time: O(n^(k-1)) in the worst case
# Space: O(k) recursion stack + output storage
def k_sum(nums, target, k):
    nums.sort()
    return _k_sum_recursive(nums, target, k, 0)


# Time: O(n^(k-1)) in the worst case
# Space: O(k) recursion stack + output storage
def _k_sum_recursive(nums, target, k, start):
    result = []
    n = len(nums)

    # Base case: 2Sum solved by two pointers.
    if k == 2:
        left, right = start, n - 1

        while left < right:
            current = nums[left] + nums[right]

            if current == target:
                result.append([nums[left], nums[right]])

                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1

                left += 1
                right -= 1
            elif current < target:
                left += 1
            else:
                right -= 1

        return result

    for i in range(start, n - k + 1):
        if i > start and nums[i] == nums[i - 1]:
            continue

        # Pruning: smallest possible already too large.
        if nums[i] * k > target:
            break

        # Pruning: even largest possible is too small.
        if nums[-1] * k < target:
            break

        subsets = _k_sum_recursive(nums, target - nums[i], k - 1, i + 1)
        for subset in subsets:
            result.append([nums[i]] + subset)

    return result
```

## Fill indices with even odd 

```python
# Problem: Sort Array by Parity II (Index Parity Constraint)
# Rearrange array so that:
# - even index contains even number
# - odd index contains odd number
#
# Example:
# Input: arr = [4,2,5,7]
# Output: [4,5,2,7]  (any valid arrangement is accepted)
#
# Constraints:
# len(arr) is even
# half of numbers are even and half are odd
# 2 <= len(arr) <= 2 * 10^4

# Time: O(n)
# Space: O(1)
def arrange_nums(arr):
    even_index = 0
    odd_index = 1
    n = len(arr)

    while even_index < n and odd_index < n:
        if arr[even_index] % 2 == 0:
            even_index += 2
        elif arr[odd_index] % 2 == 1:
            odd_index += 2
        else:
            arr[even_index], arr[odd_index] = arr[odd_index], arr[even_index]
            even_index += 2
            odd_index += 2

    return arr
```

## Reverse vowels

```python
# Problem: Reverse Vowels of a String
# Reverse only the vowels in a string and return resulting string.
#
# Example 1:
# Input: s = "hello"
# Output: "holle"
#
# Example 2:
# Input: s = "leetcode"
# Output: "leotcede"
#
# Constraints:
# 1 <= len(s) <= 3 * 10^5
# s consists of printable ASCII characters

# Time: O(n)
# Space: O(n) due to string-to-list conversion
def reverse_vowels(s):
    vowels = set("aeiouAEIOU")
    chars = list(s)

    left = 0
    right = len(chars) - 1

    while left < right:
        while left < right and chars[left] not in vowels:
            left += 1
        while left < right and chars[right] not in vowels:
            right -= 1

        chars[left], chars[right] = chars[right], chars[left]
        left += 1
        right -= 1

    return "".join(chars)
```

## Recurrence Relation : Multiplyign with indices 

```python


# Input: nums = [4,3,2,6]
# Output: 26
# Explanation:
# F(0) = (0 * 4) + (1 * 3) + (2 * 2) + (3 * 6) = 0 + 3 + 4 + 18 = 25
# F(1) = (0 * 6) + (1 * 4) + (2 * 3) + (3 * 2) = 0 + 4 + 6 + 6 = 16
# F(2) = (0 * 2) + (1 * 6) + (2 * 4) + (3 * 3) = 0 + 6 + 8 + 9 = 23
# F(3) = (0 * 3) + (1 * 2) + (2 * 6) + (3 * 4) = 0 + 2 + 12 + 12 = 26
# So the maximum value of F(0), F(1), F(2), F(3) is F(3) = 26.

# just a rotate one step clockwise function
def rotate(nums):
    for i in range(1,len(nums)):
        nums[0],nums[i] = nums[i],nums[0]
        # print(nums)
    return nums


# get all the rotations for a given array
# if n elements there are n-1 rotations are possible

array = [4,3,2,6]
for k in range(len(array)):
    print(k,array)
    rotate(array)


array = [4,8,2,6,9]
for k in range(len(array)):
    function_value = sum([i*array[i] for i in range(len(array))])
    print(k,array,function_value)
    rotate(array)


def maxRotateFunction(nums):
    n = len(nums)
    total_sum = sum(nums)
    f = sum([i * nums[i] for i in range(n)])
    ans = f
    for k in range(1, n):
        f = f + total_sum - n * nums[n - k]
        ans = max(ans, f)

    return ans

maxRotateFunction([4,8,2,6,9])

```
