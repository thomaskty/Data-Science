**Central limit theorem** : Regardless of the shape of the original population distribution, the distribution of sample means becomes approximately normal as the sample size gets large enough. $\bar X \approx N\left(\mu,\frac{\sigma^2}{n}\right)$

| Null Hypothesis (H₀) is False | Reject H₀ | Correct Decision | Power / True Positive | $1-\beta$ |

**Confidence Intervals** If we repeatedly sampled data and built intervals using the same procedure, 95% of those intervals would contain the true parameter. Computed using ; $\hat{\theta} \pm 1.96 \times SE$

**p-value** : Probability of observing the data you obtained, or something more extreme, assuming the null hypothesis is true.
