```python
def max_product(arr):
    n = len(arr)
    max_value = arr[0]
    for i in range(len(arr)):
        for j in range(i,len(arr)):
            product = arr[i]*arr[j%n]*arr[(j+1)%n]
            print([arr[i],arr[(i+1)%n],arr[(i+2)%n]],product)
            if product >max_value:
                max_value = product
            else:
                pass
    return max_value
```
