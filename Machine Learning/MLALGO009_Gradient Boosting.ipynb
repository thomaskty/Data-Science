## Gradient Boosting Algorithm

**Input:** Data $\{(x_i, y_i)\}_{i=1}^n$, differentiable loss $L(y_i, F(x))$

**Step 1:** $F_0(x) = \underset{\gamma}{\text{argmin}} \sum_{i=1}^n L(y_i, \gamma)$

**Step 2:** for $m = 1$ to $M$:
- **(A)** $r_{im} = -\left[\frac{\partial L(y_i, F(x_i))}{\partial F(x_i)}\right]_{F(x)=F_{m-1}(x)}$, $\quad i=1,\dots,n$
- **(B)** Fit regression tree to $r_{im}$ → terminal regions $R_{jm}$, $j=1\dots J_m$
- **(C)** $\gamma_{jm} = \underset{\gamma}{\text{argmin}} \sum_{x_i \in R_{ij}} L(y_i, F_{m-1}(x_i) + \gamma)$
- **(D)** $F_m(x) = F_{m-1}(x) + \nu \sum_{j=1}^{J_m} \gamma_{jm} I(x \in R_{jm})$

**Step 3:** Output $F_M(x)$
