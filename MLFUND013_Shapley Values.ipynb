# Shapley Values


- Shapley values were introduced by Lloyd Shapley (1953) in cooperative game theory. The problem:
A group of players forms a coalition and earns a total “payout.” How do we fairly attribute the payout to each player?
- Let v(S) is the payout of coalition S, then for a player $i$, the shapely value can be computed as follows ; 

$$
\LARGE \phi_i(v) = 
\sum_{S \subseteq N \setminus \{i\}}
\frac{|S|!\,(n - |S| - 1)!}{n!}
\left( v(S \cup \{i\}) - v(S) \right)
$$

 - Consider all possible orderings/permutations of players.
 - For each ordering:
    - Look at the marginal contribution of player i when they join.
    - Average the marginal contribution over all permutations.
    - Weight : How often does this combination occur in random orderings?
 

## Shapley Axioms
 - Efficiency – All the credit must be fully distributed among the players; nothing is left undistributed.
 - Symmetry – If two players contribute in exactly the same way, they must receive the same credit.
 - Null Player – If a player adds no value to any group they join, they should receive zero credit.
 - Additivity – If you combine two games, each player’s credit in the combined game should be the sum of their credits from the individual games.

## Machine Learning Application

 - Players	: Input features
 - Value function v	: Model prediction f
 - Coalition : Subset of features provided to the model
 - Payout : Expected prediction
 - Each feature “contributes” to the prediction by how much it changes the expected output when added to a subset of other features.
 - SHAP approximates (or exactly computes) the Shapley values of features for the prediction $f(x)$
 - If we look at this specific row, how much did each feature contribute to the model output?

## Consistent method for feature attribution

When you compute SHAP values for an instance \(x\):

**Base value** = expected prediction over the dataset

$$
\phi_0 = \mathbb{E}[f(X)]
$$

**Prediction decomposition:**

$$
f(x) = \phi_0 + \sum_{i=1}^{M} \phi_i
$$

Each feature’s Shapley value tells you:

- its **direction** (positive means it pushes the prediction up)
- its **magnitude** (how strong the push is)
- its **interaction**, captured implicitly by averaging over all possible coalitions

This is why SHAP is considered one of the most consistent and theoretically justified methods for feature attribution.


# Variants of SHAP

| SHAP Type          | Exact? | Model Type    | Speed       | Notes                              |
| ------------------ | ------ | ------------- | ----------- | ---------------------------------- |
| **Kernel SHAP**    | Approx | Any           | Slow        | True Shapley kernel; gold standard |
| **Tree SHAP**      | Exact  | Trees         | Very Fast   | Best for XGBoost/LightGBM          |
| **Linear SHAP**    | Exact  | Linear models | Instant     | Closed-form solution               |
| **Deep SHAP**      | Approx | Neural nets   | Medium/Fast | Uses DeepLIFT + SHAP axioms        |
| **Partition SHAP** | Approx | Any           | Fast        | Handles correlation via grouping   |
