### Sigmoid Layer
 * Applies the sigmoid function $\frac{1}{1+\exp^{-x}}$element wise 

```python
import torch
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = torch.nn.Sigmoid()
input = torch.randn(2)
output = m(input)
print(input) 
print(output)
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
sigmoid(-1.7534)
```

```python
sigmoid(0.6033)
```

```python
x_values = np.linspace(-4,4,100)
y_values = [sigmoid(j) for j in x_values]
plt.plot(x_values,y_values)
plt.grid(True)
plt.yticks(np.arange(0,1.2,0.1))
plt.title('Sigmoid Function')
plt.show()
```

### The Exponential Function
 * Negative x: Large negative values are compressed into numbers between 0 and 1.
 * Positive x: Values grow extremely quickly.
 * Compress everything from −∞ up to 0 into the finite interval (0,1], while expanding everything greater than 0 very rapidly.

```python
x_values = np.linspace(-1, 2, 100)
y_values = np.exp(x_values)

plt.plot(x_values, y_values)
plt.title('Exponential Function')

# Set y-axis ticks to integers
plt.yticks(np.arange(0, 8, 1)) 

plt.grid(True)
plt.show()
```

### Log Function 
 * Numbers between 0 and 1 are stretched into negative values.
 * Large positive numbers are compressed.
 * The logarithm maps the bounded interval (0,1] back onto the entire negative half-line (−∞,0], while compressing large positive values.

```python
x_values = np.linspace(0.02,10,100)
y_values = [np.log(j) for j in x_values]
plt.plot(x_values,y_values)
plt.xticks(np.arange(0, 10, 1)) 
plt.grid(True)
plt.title('Log Function')
plt.show()
```
