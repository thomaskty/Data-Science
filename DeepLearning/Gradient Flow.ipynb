```python
import torch
import torch.nn as nn

torch.manual_seed(42)

class DeepNetwork(nn.Module):
    def __init__(self):
        super().__init__()

        self.layers = nn.ModuleList()

        input_dim = 10
        hidden_dim = 10

        # 10 hidden layers
        for i in range(10):
            self.layers.append(nn.Linear(input_dim, hidden_dim))
            input_dim = hidden_dim

        self.output = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        self.sigmoid_outputs = []
        for layer in self.layers:
            x = torch.sigmoid(layer(x))
            self.sigmoid_outputs.append(x)

        return self.output(x)


model = DeepNetwork()

# Random input
x = torch.randn(32, 10)

# Random target
target = torch.randn(32, 1)

# Forward
prediction = model(x)

loss = ((prediction - target) ** 2).mean()

# Backward
loss.backward()

print("\nAverage Sigmoid Derivative")

print("-" * 30)

for i, activation in enumerate(model.sigmoid_outputs):

    derivative = activation * (1 - activation)

    print(

        f"Layer {i+1:2d} : {derivative.mean().item():.4f}"

    )

print(f"Loss : {loss.item():.4f}")

print("\nGradient Norms")
print("-" * 30)

for i, layer in enumerate(model.layers):
    grad = layer.weight.grad.norm().item()
    print(f"Layer {i+1:2d} : {grad:.8f}")
```

```python
import torch
import torch.nn as nn

torch.manual_seed(42)

class DeepNetwork(nn.Module):

    def __init__(self):

        super().__init__()

        self.layers = nn.ModuleList()

        input_dim = 10
        hidden_dim = 10

        for _ in range(10):

            self.layers.append(nn.Linear(input_dim, hidden_dim))
            input_dim = hidden_dim

        self.output = nn.Linear(hidden_dim, 1)

    def forward(self, x):

        for layer in self.layers:
            x = torch.sigmoid(layer(x))

        return self.output(x)


model = DeepNetwork()

# ---------------------------------------
# Intentionally make weights very large
# ---------------------------------------
for layer in model.layers:
    layer.weight.data *= 5

model.output.weight.data *= 5

x = torch.randn(32,10)
target = torch.randn(32,1)

prediction = model(x)

loss = ((prediction-target)**2).mean()

loss.backward()

print(f"Loss : {loss.item():.4f}")

print("\nGradient Norms")
print("-"*30)

for i,layer in enumerate(model.layers):

    print(
        f"Layer {i+1:2d} : {layer.weight.grad.norm().item():.6f}"
    )
```

```python
import torch
import torch.nn as nn

torch.manual_seed(42)

class DeepNet(nn.Module):

    def __init__(self):
        super().__init__()

        self.layers = nn.ModuleList()

        input_dim = 100
        hidden_dim = 100

        for _ in range(8):
            self.layers.append(nn.Linear(input_dim, hidden_dim))
            input_dim = hidden_dim

        self.output = nn.Linear(hidden_dim, 1)

        self.activations = []

    def forward(self, x):

        self.activations = []

        for layer in self.layers:

            x = torch.relu(layer(x))

            self.activations.append(x)

        return self.output(x)
```

```python
def initialize(model, method):

    for layer in model.layers:

        if method == "random":

            nn.init.normal_(layer.weight, mean=0, std=1)

        elif method == "xavier":

            nn.init.xavier_normal_(layer.weight)

        elif method == "he":

            nn.init.kaiming_normal_(layer.weight)

        nn.init.zeros_(layer.bias)

    nn.init.zeros_(model.output.bias)
```

```python
def run_experiment(method):

    model = DeepNet()

    initialize(model, method)

    x = torch.randn(64,100)

    target = torch.randn(64,1)

    prediction = model(x)

    loss = ((prediction-target)**2).mean()

    loss.backward()

    print(f"\n===== {method.upper()} =====")

    print("\nActivation Variance")

    for i,a in enumerate(model.activations):

        print(
            f"Layer {i+1}: {a.var().item():.4f}"
        )

    print("\nGradient Norm")

    for i,layer in enumerate(model.layers):

        print(
            f"Layer {i+1}: {layer.weight.grad.norm().item():.4f}"
        )
```

```python
run_experiment("random")

run_experiment("xavier")

run_experiment("he")
```
